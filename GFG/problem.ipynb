Pythagorean Triplet

```python
class Solution:
    def pythagoreanTriplet(self, arr):
        n = len(arr)
        if n < 3:
            return False
        sq = {x*x for x in arr}
        for i in range(n):
            for j in range(i+1, n):
                c = arr[i]*arr[i] + arr[j]*arr[j]
                if c in sq:
                    return True
        return False

print(Solution().pythagoreanTriplet([3, 1, 4, 6, 5]))
```

Largest number in one swap

```python
# 'Largest number in one swap'
# 'Time complexity = O(n) space complexity = O(n)'
class Solution:
    def largestSwap1(self, s):
        last = {int(s[i]): i for i in range(len(s))}
        s = list(s)

        for i, ch in enumerate(s):
            cur = int(ch)

            for d in range(9, cur, -1):
                if d in last and last[d] > i:
                    j = last[d]
                    s[i], s[j] = s[j], s[i]
                    return ''.join(s)

        return ''.join(s)
# 'Optimal O(n) space O(1) solution'
class Solution:
    def largestSwap(self, s):
            s=list(s)
            n=len(s)
            max_ind=n-1
            l=r=-1
            for i in range(n-1,-1,-1):
                if s[i]>s[max_ind]:
                    max_ind=i
                elif s[i]<s[max_ind]:
                    l=i
                    r=max_ind
            if l!=-1:
                s[l],s[r]=s[r],s[l]
            return "".join(s)

print(Solution().largestSwap("2736"))
```

Union of 2 sorted arrays

```python
#Brute
def findUnion(a, b):
    s=set()
    for i in a:
        s.add(i)
    for i in b:
        s.add(i)
    return sorted(s)
#optimal Two pinters approach
def findUnion1(a, b):
    n1=len(a)
    n2=len(b)
    i=j=0
    union_arr=[]
    while i<n1 and j<n2:
        if a[i]<=b[j]:
            if len(union_arr)==0 or union_arr[-1] != a[i]:
                union_arr.append(a[i])
            i+=1
        else:
            if len(union_arr)==0 or union_arr[-1] != b[j]:
                union_arr.append(b[j])
            j+=1
    while j<n2:
        if len(union_arr)==0 or union_arr[-1] !=b[j]:
            union_arr.append(b[j])
            j+=1
    while i<n1:
        if len(union_arr)==0 or union_arr[-1] !=a[i]:
            union_arr.append(a[i])
            i+=1
    return union_arr


a = [1, 2, 3, 4, 5]
b = [1, 2, 3, 6, 7]
print(findUnion1(a,b))
```

Array Leaders

```python
#Brute force approach                           # Time complexity ~= O(n^2) and Space complexity: O(n)
def leaders(arr):
    n = len(arr)
    res = []
    for i in range(n):
        leader = True
        for j in range(i+1,n):
            if arr[j] > arr[i]:
                leader = False
        if leader:
            res.append(arr[i])
    return res

#Optimal approach
```
