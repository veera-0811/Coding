```python
def pat1(n):
    for i in range(n):
        for j in range(n):
            print("*",end="")
        print()
def pat2(n):
    for i in range(n):
        for j in range(0,i+1):
            print("*",end="")
        print()
def pat3(n):
    for i in range(1,n+1):
        for j in range(1,i+1):
            print(j,end=" ")
        print()
def pat4(n):
    for i in range(1,n+1):
        for j in range(1,i+1):
            print(i,end="")
        print()
def pat5(n):
    for i in range(n):
        for j in range(n-i):
            print("*",end="")
        print()
def pat6(n):
    for i in range(1,n+1):
        for j in range(1,n-i+2):
            print(j,end="")
        print()
def pat7(n):
    for i in range(n):
        #space
        for j in range(n-i-1):
            print(" ",end="")
        #star
        for j in range(2*i+1):
            print("*",end="")
        #space
        for j in range(n-i-1):
            print(" ",end="")
        print()
def pat8(n):
    for i in range(n):
        #space
        for j in range(i):
            print(" ",end="")
        #star
        for j in range(2*n-(2*i+1)):
            print("*",end="")
        #space
        for j in range(i):
            print(" ",end="")
        print()
def pat9(n):
    for i in range(1,2*n):
        stars=i
        if i>n:
            stars=2*n-i
        for j in range(1,stars+1):
            print("*",end="")
        print()

n=int(input("Enter the number:"))

pat5(n)
print()
pat6(n)
print()
pat7(n)
print()
pat8(n)
print()
pat9(n)
```

```python
#GCD of two numbers
def gcd(a,b):
    while a>0 and b>0:
        if a>b:
            a=a%b
        else:
            b=b%a
    if a==0:
        return b
    else:
        return a
a,b=map(int,input().split())
print(f"GCD of {a} and {b} is",gcd(a,b))
```

```python
#longest balanced binary substring
s=input().strip()
pf=0
max_len=0
seen={0:-1}
for i in range(len(s)):
    if s[i]=='1':
        pf+=1
    else:
        pf-=1
    if pf in seen:
        length=i-seen[pf]
        max_len=max(max_len,length)
    else:
        seen[pf]=i
print(max_len)
```

```python
#remove duplicates from an array
def dup(a):
    i=0
    for j in range(1,len(a)):
        if a[i]!=a[j]:
            a[i+1]=a[j]
            i+=1
    return i+1

nums=[1,2,2,3]
print(dup(nums))
```

```python
# left rotate array by d places
#Brute
def lrd(arr,d):
    n=len(arr)
    d=d%n
    temp=[0]*(d)
    for i in range(d):
        temp[i]=arr[i]
    for i in range(d,n):
        arr[i-d]=arr[i]
    for i in range(n-d,n):
        arr[i]=temp[i-(n-d)]
    return arr

#optimal
def reverse(arr,l,r):
    while l<r:
        arr[l],arr[r]=arr[r],arr[l]
        l+=1
        r-=1
def lrd1(arr,d):
    n=len(arr)
    d=d%n

    reverse(arr,0,d-1)
    reverse(arr,d,n-1)
    reverse(arr,0,n-1)
    return arr

arr=list(map(int,input().split()))
d=int(input())
print(arr,d)
print(lrd1(arr,d))
```

```python
# right rotate array by d places
class Solution:
    def rotate(self, nums: List[int], k: int) -> None:
        # Do not return anything, modify nums in-place instead.
        def reverse(arr,l,r):
            while l<r:
                arr[l],arr[r]=arr[r],arr[l]
                l+=1
                r-=1
        n=len(nums)
        k=k%n
        reverse(nums,n-k,n-1)
        reverse(nums,0,n-k-1)
        reverse(nums,0,n-1)
        return nums
    
nums = [1,2,3,4,5,6,7]
k = 3
print(Solution().rotate(nums,k))
```

```python
#Brute
def findUnion(a, b):
    s=set()
    for i in a:
        s.add(i)
    for i in b:
        s.add(i)
    return sorted(s)
#optimal Two pointers approach
def findUnion1(a, b):
    n1=len(a)
    n2=len(b)
    i=j=0
    union_arr=[]
    while i<n1 and j<n2:
        if a[i]<=b[j]:
            if len(union_arr)==0 or union_arr[-1] != a[i]:
                union_arr.append(a[i])
            i+=1
        else:
            if len(union_arr)==0 or union_arr[-1] != b[j]:
                union_arr.append(b[j])
            j+=1
    while j<n2:
        if len(union_arr)==0 or union_arr[-1] !=b[j]:
            union_arr.append(b[j])
            j+=1
    while i<n1:
        if len(union_arr)==0 or union_arr[-1] !=a[i]:
            union_arr.append(a[i])
            i+=1
    return union_arr


a = [1, 2, 3, 4, 5]
b = [1, 2, 3, 6, 7]
print(findUnion1(a,b))
```
