GFG -> BFS of graph & DFS of Graph

```python
class Solution:
    def bfs(self, adj):
        res = []
        visited = [False]*len(adj)
        q = [0]
        visited[0] = True
        while q:
            node = q.pop(0)
            res.append(node)
            for neighbour in adj[node]:
                if not visited[neighbour]:
                    q.append(neighbour)
                    visited[neighbour] = True
        return res
    
    def dfs(self, adj):
        res = []
        vis = [False]*len(adj)
        def helper(ver):
            if vis[ver]:
                return
            res.append(ver)
            vis[ver] = True
            for neigh in adj[ver]:
                helper(neigh)
        helper(0)
        return res
    
adj = [[2, 3, 1], [0], [0, 4], [0], [2]]
print(Solution().bfs(adj))
print(Solution().dfs(adj))
```

GFG Topological Sort {Kahn’s Algorithm}

```python
#Topological sort works when:
#You have a directed acyclic graph (DAG)
#There are fixed dependencies like “A must happen before B”
class Solution:
    def topoSort(self, V, edges):
        adj = [[] for i in range(V)]
        indeg = [0]*V
        res = []
        for u,v in edges:
            adj[u].append(v)
            indeg[v] += 1
            
        q = []
        for i in range(V):
            if indeg[i] == 0:
                q.append(i)
        while q:
            node = q.pop(0)
            res.append(node)
            for neigh in adj[node]:
                indeg[neigh] -= 1
                if indeg[neigh] == 0:
                    q.append(neigh)
        return res
V = 4
edges = [[3, 0], [1, 0], [2, 0]]
print(Solution().topoSort(V, edges))
```

LEETCODE 207. Course Schedule


```python
class Solution:
    def canFinish(self, numCourses: int, prerequisites: list[list[int]]) -> bool:
        indeg = [0]*numCourses
        adj = [[] for _ in range(numCourses)]
        for u,v in prerequisites:
            adj[v].append(u)
            indeg[u] += 1
        q = []
        completed = 0
        for i in range(numCourses):
            if indeg[i] == 0:
                q.append(i)
                completed += 1
        if completed == 0:
            return False
        while q:
            node = q.pop(0)
            for neigh in adj[node]:
                indeg[neigh] -= 1
                if indeg[neigh] == 0:
                    q.append(neigh)
                    completed += 1
        return completed == numCourses
```

LEETCODE 994. Rotting Oranges

```python
from collections import deque
class Solution:
    def orangesRotting(self, grid):
        rows, cols = len(grid), len(grid[0])
        q = deque()
        fresh = 0

        # Step 1: Initialize
        for r in range(rows):
            for c in range(cols):
                if grid[r][c] == 2:
                    q.append((r, c))
                elif grid[r][c] == 1:
                    fresh += 1

        # Directions: up, down, left, right
        directions = [(1,0), (-1,0), (0,1), (0,-1)]
        time = 0

        # Step 2: BFS
        while q and fresh > 0:
            for _ in range(len(q)):
                r, c = q.popleft()
                for dr, dc in directions:
                    nr = r + dr
                    nc = c + dc

                    if 0 <= nr < rows and 0 <= nc < cols and grid[nr][nc] == 1:
                        grid[nr][nc] = 2
                        fresh -= 1
                        q.append((nr, nc))
            time += 1

        # Step 3: Result
        return time if fresh == 0 else -1
    
grid = [[2,1,1],[1,1,0],[0,1,1]]
print(Solution().orangesRotting(grid))
```

LEETCODE 200. Number of Islands

```python
class Solution:
    def numIslands(self, grid: list[list[str]]) -> int:
        m,n = len(grid),len(grid[0])
        def dfs(row,col):
            if row>=m or row<0 or col>=n or col<0 or grid[row][col] == "0":
                return
            grid[row][col] = "0"
            dfs(row+1,col)
            dfs(row-1,col)
            dfs(row,col+1)
            dfs(row,col-1)

        islands = 0
        for i in range(m):
            for j in range(n):
                if grid[i][j] == "1":
                    islands += 1
                    dfs(i,j)
        return islands
```

LEETCODE 695. Max Area of Island

```python
class Solution:
    def maxAreaOfIsland(self, grid: list[list[int]]) -> int:
        m,n = len(grid),len(grid[0])
        def dfs(row,col):
            if row>=m or row<0 or col>=n or col<0 or grid[row][col] == 0:
                return 0
            grid[row][col] = 0
            return(1+dfs(row+1,col)+
            dfs(row-1,col)+
            dfs(row,col+1)+
            dfs(row,col-1))

        maxarea = 0
        for i in range(m):
            for j in range(n):
                if grid[i][j] == 1:
                    area = dfs(i,j)
                    maxarea = max(area,maxarea)
        return maxarea
```

LEETCODE 1971. Find if Path Exists in Graph

```python
# BFS Approach
class Solution:
    def validPath(self, n: int, edges: list[list[int]], source: int, destination: int) -> bool:
        graph = [[] for _ in range(n)]
        for u,v in edges:
            graph[u].append(v)            
            graph[v].append(u)      
        visited = [False]*n      
        q = deque([source])
        visited[source] = True
        while q:
            node = q.popleft()
            if node == destination:
                return True
            
            for neigh in graph[node]:
                if not visited[neigh]:
                    visited[neigh] = True
                    q.append(neigh)
        return False
```

GFG UnionFind

```python
class Solution:
    def find(self,par,x):
        if x != par[x]:
            return self.find(par,par[x])
        return x
    
    #Function to merge two nodes a and b.
    def union_(self,a,b,par,rank1):
        pa = self.find(par,a)
        pb = self.find(par,b)
        if rank1[pa] < rank1[pb]:
            par[pa] = pb
        elif rank1[pb] < rank1[pa]:
            par[pb] = pa
        else:
            par[pa] = pb
            rank1[pb] += 1
        
    #Function to check whether 2 nodes are connected or not.
    def isConnected(self,x,y,par,rank1):
        return self.find(par,x) == self.find(par,y)
```

LEETCODE 547. Number of Provinces

```python
#UnionFind Approach
class Solution:
    def findCircleNum(self, isConnected: list[list[int]]) -> int:
        n = len(isConnected)
        par = [i for i in range(n)]
        def find(x):
            if x != par[x]:
                return find(par[x])
            return x
        def union(a,b):
            pa = find(a)
            pb = find(b)
            par[pa] = pb

        for i in range(n):
            for j in range(n):
                if isConnected[i][j] == 1:
                    union(i,j)
        res = set()
        for i in range(n):
            res.add(find(i))
        return len(res)
# BFS Approach
```

GFG Connected Components in an Undirected Graph


```python
# UnionFind Approach
from collections import defaultdict
class Solution:
    def getComponents(self, V, edges):
        parent = [i for i in range(V)]
        rank = [0]*V
        def find(x):
            if parent[x] != x:
                return find(parent[x])
            return x
        def union(a,b):
            pa = find(a)
            pb = find(b)
            if rank[pa]<rank[pb]:
                parent[pa]=pb
            elif rank[pb]<rank[pa]:
                parent[pb]=pa
            else:
                parent[pa] = pb
                rank[pb] += 1
        
        for u,v in edges:
            union(u,v)
        d = defaultdict(list)
        for i in range(V):
            p = find(i)
            d[p].append(i)
            
        return list(d.values())
    
V = 5
edges = [[0, 1], [2, 1], [3, 4]]
print(Solution().getComponents(V, edges))
```

GFG 684. Redundant Connection

```python
# UnionFind Approach
class Solution:
    def findRedundantConnection(self, edges: list[list[int]]) -> list[int]:
        n = len(edges)
        parent = [i for i in range(n+1)]
        def find(x):
            if parent[x] != x:
                return find(parent[x])
            return x
        def union(a,b):
            pa = find(a)
            pb = find(b)
            parent[pa] = pb
        for u,v in edges:
            if find(u) == find(v):
                return [u,v]
            union(u,v)
```
