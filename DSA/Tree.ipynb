```python
#LEETCODE 100 SAME TREE
# Definition for a binary tree node.
class TreeNode(object):
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


# BFS Approach
from collections import deque
class Solution(object):
    def isSameTree(self, p, q):
        a=deque([(p,q)])
        while a:
            node1,node2=a.popleft()
            if not node1 and not node2:
                continue
            if not node1 or not node2:
                return False
            if node1.val!=node2.val:
                return False
            a.append((node1.left,node2.left))
            a.append((node1.right,node2.right))
        return True

# RECURSIVE APPROACH
class Solution:
    def isSameTree1(self,p,q):
        if not p and not q:
            return True
        if not p or not q or p.val!=q.val:
            return False
        return self.isSameTree1(p.left,q.left) and self.isSameTree1(p.right,q.right)
```

```python
# LEETCODE 226 INVERT BINARY TREE 

# BFS Approach
from collections import deque
class Solution:
    def invertTree(self, root):
        if not root:
            return None
        q=deque([root])
        while q:
            node=q.popleft()
            node.left,node.right=node.right,node.left
            if node.left:
                q.append(node.left)
            if node.right:
                q.append(node.right)
        return root
# OR RECURSION
class Solution:
    def invertTree(self, root):
        if not root:
            return None
        root.left,root.right=root.right,root.left
        self.invertTree(root.left)
        self.invertTree(root.right)
        return root
```

LC 101 SYMMETRIC TREE

```python
# BFS Approach
from collections import deque
class Solution:
    def isSymmetric(self,root):
        if not root:
            return True
        q=deque([(root.left,root.right)])
        while q:
            node1,node2=q.popleft()
            if not node1 and not node2:
                continue
            if not node1 or not node2:
                return False
            if node1.val!=node2.val:
                return False
            q.append((node1.left,node2.right))
            q.append((node1.right,node2.left))
        return True
# OR RECURSION
class Solution:
    def isSymmetric(self,root):
        def isMirror(n1,n2):
            if not n1 and not n2:
                return True
            if not n1 or not n2 or n1.val != n2.val:
                return False
            return isMirror(n1.left,n2.right) and isMirror(n1.right,n2.left)
        
        return isMirror(root,root)
```

LEETCODE 102 Binary Tree Level Order Traversal

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right

#Here, we are doing a level order traversal and storing the values of each level in a list. We use a queue to keep track of the nodes at each level. 
# We pop nodes from the queue, add their values to the current level list, and then add their left and right children to the queue for the next level. 
# Finally, we return the list of levels.

class Solution:
    def levelOrder(self, root: Optional[TreeNode]) -> List[List[int]]:
        if not root:
            return []
        q = [root]
        res = []
        while q:
            lev = []
            n = len(q)
            for i in range(n):
                node = q.pop(0)
                lev.append(node.val)
                if node.left:
                    q.append(node.left)
                if node.right:
                    q.append(node.right)
            res.append(lev)
        return res
```

LEETCODE 94 Binary Tree In Order Traversal

```python
class Solution:
    def inorderTraversal(self, root: Optional[TreeNode]) -> List[int]:
        res = []
        def helper(root,res):
            if not root:
                return
            helper(root.left,res)
            res.append(root.val)
            helper(root.right,res)
        helper(root,res)
        return res
```

LEETCODE 653. Two Sum IV - Input is a BST

```python
# Here, we are doing an inorder traversal and storing the values in a list. 
# Then we are using two pointers to check if there are two numbers in the list that add up to k. 
# If we find such a pair, we return True. If we exhaust the list without finding such a pair, we return False.

class Solution:
    def findTarget(self, root: Optional[TreeNode], k: int) -> bool:
        res = []
        def helper(root,res):
            if not root:
                return
            helper(root.left,res)
            res.append(root.val)
            helper(root.right,res)
        helper(root,res)
        l,r = 0,len(res)-1
        while l < r:
            if res[l] + res[r] == k:
                return True
            elif res[l] + res[r] < k:
                l += 1
            else:
                r -= 1
        return False
```

LEETCODE 112 Path Sum

```python
# Here, we are doing a depth first search and keeping track of the current path and the current sum. 
# When we reach a leaf node, we check if the current sum is equal to the target sum. If it is, we add the current path to the result list.
class Solution:
    def hasPathSum(self, root: Optional[TreeNode], targetSum: int) -> bool:
        if not root:
            return False
        if not root.left and not root.right:
            return targetSum==root.val
        return self.hasPathSum(root.left,targetSum-root.val) or self.hasPathSum(root.right,targetSum-root.val)
```

LEETCODE 113 Path Sum II

```python
# Here, we are doing a depth first search and keeping track of the current path and the current sum. 
# When we reach a leaf node, we check if the current sum is equal to the target sum. If it is, we add the current path to the result list.
class Solution:
    def pathSum(self, root: Optional[TreeNode], targetSum: int) -> List[List[int]]:
        res = []
        def helper(root,path,cur):
            if not root:
                return
            cur += root.val
            path.append(root.val)
            if not root.left and not root.right and cur==targetSum:
                res.append(path[:])
            helper(root.left,path,cur)
            helper(root.right,path,cur)
            path.pop()
        helper(root,[],0)
        return res
```

LEETCODE 114. Flatten Binary Tree to Linked List

```python
# Here, we are doing a preorder traversal and storing the nodes in a list. 
# Then we are iterating through the list and changing the left and right pointers of each node to flatten the tree into a linked list.
class Solution:
    def flatten(self, root: Optional[TreeNode]) -> None:
        res = []
        def helper(root,res):
            if not root:
                return
            res.append(root)
            helper(root.left,res)
            helper(root.right,res)
        helper(root,res)
        for node in res[1:]:
            root.left = None
            root.right = node
            root = root.right
#                                                   OR
# Here, we are doing a reverse preorder traversal and keeping track of the previous node. 
# We first flatten the right subtree, then the left subtree, and then we set the current node's right pointer to the previous node and the left pointer to None.
class Solution:
    def __init__(self):
        self.prev = None
    def flatten(self, root: Optional[TreeNode]) -> None:
        if not root:
            return None
        self.flatten(root.right)
        self.flatten(root.left)

        root.right = self.prev
        root.left = None
        
        self.prev = root
```

LEETCODE 297. Serialize and Deserialize Binary Tree

```python
# Definition for a binary tree node.
class TreeNode(object):
    def __init__(self, x):
        self.val = x
        self.left = None
        self.right = None

class Codec:
    def serialize(self, root):
        if not root:
            return ""
        res = []
        q = [root]
        while q:
            node = q.pop(0)
            if node:
                res.append(str(node.val))
                q.append(node.left)
                q.append(node.right)
            else:
                res.append("#")
        return ",".join(res)

    def deserialize(self, data):
        if not data:
            return None
        nodes = data.split(",")
        root = TreeNode(int(nodes[0]))
        q = [root]
        i = 1
        while q:
            node = q.pop(0)
            if nodes[i] != "#":
                node.left = TreeNode(int(nodes[i]))
                q.append(node.left)
            i += 1
            if nodes[i] != "#":
                node.right = TreeNode(int(nodes[i]))
                q.append(node.right)
            i += 1
        return root

# Your Codec object will be instantiated and called as such:
# ser = Codec()
# deser = Codec()
# ans = deser.deserialize(ser.serialize(root))
```

LEETCODE 230. Kth Smallest Element in a BST

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def kthSmallest(self, root,k):
        res = []
        def helper(root,res):
            if not root:
                return
            helper(root.left,res)
            res.append(root.val)
            helper(root.right,res)
        helper(root,res)
        return res[k-1]
```

LEETCODE 700. Search in a Binary Search Tree

```python
# Definition for a binary tree node.
# class TreeNode:
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution:
    def searchBST(self, root: Optional[TreeNode], val: int) -> Optional[TreeNode]: # type: ignore
        if not root:
            return
        if val == root.val:
            return root
        elif val < root.val:
            return self.searchBST(root.left,val)
        else:
            return self.searchBST(root.right,val)    
```

LEETCODE 701. Insert into a Binary Search Tree

```python
from typing import Optional

# Recursive Approach
class Solution:
    def insertIntoBST(self, root: Optional[TreeNode], val: int) -> Optional[TreeNode]:
        if not root:
            return TreeNode(val)
        if val < root.val:
            root.left = self.insertIntoBST(root.left,val)
        else:
            root.right = self.insertIntoBST(root.right,val)
        return root
    

# Iterative Approach
class Solution:
    def insertIntoBST(self, root: Optional[TreeNode], val: int) -> Optional[TreeNode]:
        if not root:
            return TreeNode(val)
        
        curr = root
        while True:
            if val < curr.val:
                # If there's a left child, keep moving left
                if curr.left:
                    curr = curr.left
                # Otherwise, insert the new node here
                else:
                    curr.left = TreeNode(val)
                    break
            else:
                # If there's a right child, keep moving right
                if curr.right:
                    curr = curr.right
                # Otherwise, insert the new node here
                else:
                    curr.right = TreeNode(val)  
        return root
```
