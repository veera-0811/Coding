LEETODE 455 :- ASSIGN COOKIES

```python
class Solution:
    def findContentChildren(self, g: List[int], s: List[int]) -> int:
        g.sort()
        s.sort()
        i=j=0
        while i<len(g) and j<len(s):
            if s[j]>=g[i]:
                i+=1
            j+=1
        return i
```

LEETCODE 2410 :- maximum-matching-of-players-with-trainers (similar sum to above sum)

```python
class Solution:
    def matchPlayersAndTrainers(self, players: List[int], trainers: List[int]) -> int:
        players.sort()
        trainers.sort()
        i=j=0
        while i<len(players) and j<len(trainers):
            if players[i]<=trainers[j]:
                i+=1
            j+=1
        return i
```
