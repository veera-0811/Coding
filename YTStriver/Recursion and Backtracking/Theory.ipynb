Print numbers from 1 to N

```python
# print linearly from 1 to N
def func(i,n):
    if i > n:
        return
    print(i,end = " ")
    func(i+1,n)

n = 5
func(1,n)

print()

# print linearly from 1 to N but by BACKTRACKING
def func(i):
    if i < 1:
        return
    func(i-1)
    print(i,end =" ")

func(5)
    
```

Sum of N numbers

```python
# PARAMETERISED RECURSION
def f(num,sum):
    if num < 1:
        print(sum)
        return
    f(num-1,sum+num)

f(4,0)

# FUNCTIONAL RECURSION  -> IMP
def f(n):
    if n == 0:
        return 0
    return n + f(n-1)

f(4)
```

Factorial of a number

```python
# PARAMETERISED RECURSION
def fact(n,res):
    if n < 1:
        print(res)
        return
    fact(n-1,res*n)
fact(4,1)

# FUNCTIONAL RECURSION
def fact(n):
    if n == 0:
        return 1
    return n * fact(n-1)
fact(4)
```

Reverse an array

```python
arr = [1,2,3,4,5]

def rev1(arr,l,r):
    if l >= r:
        return
    arr[l],arr[r] = arr[r],arr[l]
    rev1(arr,l+1,r-1)

rev1(arr,0,len(arr)-1)
print(arr)

arr2 = [6,7,8,9,10,11]
def rev(arr2,n,i):
    if i >= n//2:
        return
    arr2[i],arr2[n-i-1] = arr2[n-i-1],arr2[i]
    rev(arr2,n,i+1)
    
rev(arr2,len(arr2),0)
print(arr2)
```

check Palindrome

```python
s = "madam"
def pal(s,n,i):
    if i >= n//2:
        return True
    if s[i] != s[n-i-1]:
        return False
    return pal(s,n,i+1)
pal(s,len(s),0)
```

Fibanocci number

```python
# Time Complexity: O(2^n)  because for every call, we are making two extra calls which was exponential in nature
# All the remaining above recursion programs are O(n)
def fib(n):
    if n <= 1:
        return n
    return fib(n-1) + fib(n-2)
fib(5)   # 5th fibanocci number is 5
```

##### SUBSEQUENCES
Suppose, There is an array [3,1,2]
- Contiguous Subsequence    :- [],[3],[1],[2],[3,1],[1,2],[3,1,2]
- Non-Contiguous Subsequence:- [3,2]

**Total subsequences = 2³ = 8**  where 3 is size of the array

GENERATE SUBSEQUENCES

```python
# Time Complexity:- O( n * 2^n)    where for printing the array takes O(n) and for every element, there is a chance of pick / Not-pick. so O(2^n)
# Auxiliary Space: O(n)
a = [3,1,2]
n = len(a)

def func(ind,res):
    if ind >= n:
        print(res)
        return
    res.append(a[ind])
    func(ind+1,res)     # Pick / Take
    res.pop()
    func(ind+1,res)     # Not Pick / Not-Take
func(0,[])
```

Generate the above subsequences in reverse order

```python
a = [3,1,2]
n = len(a)

def func(ind,res):
    if ind >= n:
        print(res)
        return
    func(ind+1,res)     # Not Pick
    res.append(a[ind])
    func(ind+1,res)     # Pick
    res.pop()
func(0,[])
```

##### ***ALL KINDS OF PATTERNS IN RECURSION***

1. Printing Subsequences whose sum is k

```python
def pos(ind,arr,n,temp,s,k):
    if ind == n:
        if s == k:
            print(temp)
        return
    temp.append(arr[ind])
    s += arr[ind]
    pos(ind+1,arr,n,temp,s,k)
    temp.pop()
    s -= arr[ind]
    pos(ind+1,arr,n,temp,s,k)

arr,k = [1,2,1],2
n = len(arr)
temp = []
pos(0,arr,n,temp,0,k)
```

2. Printing only one Subsequence whose sum is k & Avoid further Recursion calls.

```python
def pos1(ind,arr,n,temp,s,k):
    if ind == n:
        if s == k:
            print(temp)
            return True
        return False
    temp.append(arr[ind])
    s += arr[ind]
    if pos1(ind+1,arr,n,temp,s,k):
        return True
    temp.pop()
    s -= arr[ind]
    if pos1(ind+1,arr,n,temp,s,k):
        return True
    return False

arr,k = [1,2,1],2
n = len(arr)
temp = []
pos1(0,arr,n,temp,0,k)
```

3. Count the Subsequences whose sum is k

```python
def pos2(ind,arr,n,s,k):
    if s > k:               # IMP Slight Optimisation step -> This is the optimization step, if the sum is already greater than k, then we don't need to go further down the recursion tree.
        return 0            # Strictly done if array has only positive numbers, if it has negative numbers, then this optimization step will not work.

    if ind == n:
        if s == k:
            return 1
        return 0
    s += arr[ind]
    l = pos2(ind+1,arr,n,s,k)
    s -= arr[ind]
    r = pos2(ind+1,arr,n,s,k)
    return l + r

arr,k = [1,2,1],2
n = len(arr)
print("The no of subsequences with sum",k,"is",pos2(0,arr,n,0,k))
```

#### MERGE SORT (Divide and Merge Algorithm)

Approach - 1 -> Recommended

```python
def merge(arr,low,mid,high):
    temp = []
    i = low
    j = mid+1
    while i <= mid and j <= high:
        if arr[i] <= arr[j]:
            temp.append(arr[i])
            i += 1
        else:
            temp.append(arr[j])
            j += 1
    while i <= mid:
        temp.append(arr[i])
        i += 1
    while j <= high:
        temp.append(arr[j])
        j += 1

    for i in range(len(temp)):              # IMP -> This is the step where we are copying the elements from temp array to original array. This is the step which is often missed by many people. So, please remember this step. why low+i is used because we are copying the elements from temp array to original array starting from low index to high index.
        arr[low+i] = temp[i]                # OR arr[i] = temp[i-low]  -> This is also correct because we are copying the elements from temp array to original array starting from low index to high index. So, we can use either low+i or i-low. Both are correct.

def merge_sort(arr,low,high):
    if low >= high:
        return
    mid = (low+high)//2
    merge_sort(arr,low,mid)
    merge_sort(arr,mid+1,high)
    merge(arr,low,mid,high)

arr = [38, 27, 43, 3, 9, 82, 10]
merge_sort(arr,0,len(arr) - 1)
print(arr)
```

Approach - 2

```python
def merge_sort(arr):
    if len(arr) > 1:

        mid = len(arr) // 2

        left = arr[:mid]
        right = arr[mid:]

        merge_sort(left)
        merge_sort(right)

        i = j = k = 0

        while i < len(left) and j < len(right):
            if left[i] < right[j]:
                arr[k] = left[i]
                i += 1
            else:
                arr[k] = right[j]
                j += 1
            k += 1

        while i < len(left):
            arr[k] = left[i]
            i += 1
            k += 1

        while j < len(right):
            arr[k] = right[j]
            j += 1
            k += 1


arr = [38, 27, 43, 3, 9, 82, 10]
merge_sort(arr)
print(arr) 
```

#### QUICK SORT (Divide and Conquer Algorithm)

```python
def quicksort(arr,low,high):
    if low < high:
        pivot = partition(arr,low,high)
        quicksort(arr,low,pivot - 1)
        quicksort(arr,pivot + 1,high)

def partition(arr,low,high):
    pivot = arr[low]
    i = low
    j = high
    while i < j:
        while arr[i] <= pivot and i <= high:
            i += 1
        while arr[j] > pivot and j >= low:
            j -= 1
        if i < j:
            arr[i],arr[j] = arr[j],arr[i]
    arr[low],arr[j] = arr[j],arr[low]
    return j

arr = [38, 27, 43, 3, 9, 82, 10]
quicksort(arr,0,len(arr) - 1)
print(arr)
```
