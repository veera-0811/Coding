**Binary Tree:-** A non-liner tree data structure in which each node has at most two children.

Full Binary Tree:- A binary tree in which every node has either 0 or 2 children.

Complete Binary Tree:- A binary tree in which all levels are completely filled except possibly the last level, which is filled from left to right.

Perfect Binary Tree:- A binary tree in which all internal nodes have two children and all leaf nodes are at the same level.

Balanced Binary Tree:- A binary tree in which the height difference between the left and right subtrees of every node is at most 1.

Degenerate (Skewed) Binary Tree:- A binary tree in which every parent node has only one child.

Left-Skewed Binary Tree:- A binary tree in which every node has only a left child.

Right-Skewed Binary Tree:- A binary tree in which every node has only a right child.

### Binary Tree Representation

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

root = Node(10)

root.left = Node(5)
root.right = Node(20)

root.left.left = Node(2)
root.left.right = Node(8)

root.right.right = Node(30)

# print entire tree (no need)
def print_tree(node, level=0):
    if node is not None:
        print_tree(node.right, level + 1)
        print(' ' * 4 * level + '->', node.data)
        print_tree(node.left, level + 1)

print_tree(root)

#        10
#       /  \
#      5    20
#     / \     \
#    2   8     30
```

#### TAKING INPUT BINARY TREE USING QUEUE (Append level by level) DYNAMICALLY

```python
from collections import deque

# ---------------- Node Class ----------------
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

# ---------------- Build Tree ----------------
from collections import deque
def buildTree(arr):
    if len(arr) == 0 or arr[0] == "N":
        return None

    root = Node(int(arr[0]))
    queue = deque([root])

    i = 1

    while queue and i < len(arr):

        current = queue.popleft()

        # Left Child
        if i < len(arr) and arr[i] != "N":
            current.left = Node(int(arr[i]))
            queue.append(current.left)
        i += 1

        # Right Child
        if i < len(arr) and arr[i] != "N":
            current.right = Node(int(arr[i]))
            queue.append(current.right)
        i += 1

    return root

# ---------------- Inorder ----------------
def inorder(root):
    if root is None:
        return

    inorder(root.left)
    print(root.data, end=" ")
    inorder(root.right)

# ---------------- Preorder ----------------
def preorder(root):
    if root is None:
        return

    print(root.data, end=" ")
    preorder(root.left)
    preorder(root.right)

# ---------------- Postorder ----------------
def postorder(root):
    if root is None:
        return

    postorder(root.left)
    postorder(root.right)
    print(root.data, end=" ")

# ---------------- Driver Code ----------------

arr = input("Enter tree in level order: ").split()

root = buildTree(arr)

print("\nInorder Traversal:")
inorder(root)

print("\n\nPreorder Traversal:")
preorder(root)

print("\n\nPostorder Traversal:")
postorder(root)
```

#### LEVEL ORDER TRAVERSAL

```python
# ---------------- Level Order ----------------
def levelOrder(root):
    if root is None:
        return

    queue = deque([root])

    while queue:

        node = queue.popleft()

        print(node.data, end=" ")

        if node.left:
            queue.append(node.left)

        if node.right:
            queue.append(node.right)

print("\n\nLevel Order Traversal:")
levelOrder(root)
```

##### ITERATIVE **PREORDER, INORDER, POSTORDER** TRAVERSAL IN BINARY TREE

```python
def preorderTraversal(root):
    if not root:
        return []
    
    stack = [root]
    ans = []
    while stack:
        node = stack.pop()
        ans.append(node.val)
        if node.right:
            stack.append(node.right)
        if node.left:
            stack.append(node.left)
    return ans

def inorderTraversal(root):
    stack = []
    ans = []
    current = root
    while current or stack:

        while current:
            stack.append(current)
            current = current.left

        current = stack.pop()
        ans.append(current.val)
        current = current.right
    return ans

# Method 1 (Two Stacks) -> Easy (Recommended for Exam)
def postorderTraversal(root):
    if not root:
        return []

    stack1 = [root]
    stack2 = []
    while stack1:
        node = stack1.pop()
        stack2.append(node)
        if node.left:
            stack1.append(node.left)
        if node.right:
            stack1.append(node.right)

    ans = []
    while stack2:
        ans.append(stack2.pop().val)
    return ans

# Method 2 (One Stack) -> Interview Level
def postorderTraversal(root):
    if root is None:
        return []

    stack = []
    curr = root
    prev = None
    ans = []
    while curr or stack:

        # Go to the leftmost node
        if curr:
            stack.append(curr)
            curr = curr.left

        else:
            # Look at the top node
            curr = stack[-1]

            # If right child is absent
            # OR already processed
            if curr.right == None or curr.right == prev:
                ans.append(curr.val)
                stack.pop()
                prev = curr
                curr = None
            else:
                # Visit right subtree
                curr = curr.right
    return ans
```

####  Preorder Inorder Postorder Traversals in **One Traversal**

```python
def allTraversals(root):

    if root is None:
        return [], [], []

    preorder = []
    inorder = []
    postorder = []

    stack = [(root, 1)]

    while stack:

        node, state = stack.pop()

        # ---------- Preorder ----------
        if state == 1:

            preorder.append(node.val)

            stack.append((node, 2))

            if node.left:
                stack.append((node.left, 1))

        # ---------- Inorder ----------
        elif state == 2:

            inorder.append(node.val)

            stack.append((node, 3))

            if node.right:
                stack.append((node.right, 1))

        # ---------- Postorder ----------
        else:

            postorder.append(node.val)

    return preorder, inorder, postorder
```

Interview Tip

Whenever you need a recursive helper function to update one value globally (like diameter, maximum path sum, or longest path), there are three common Python approaches:

* Use a class variable (self.diameter) -> (most common)
* Pass a mutable list (diameter = [0]) -> (equivalent to the Java array approach)
* Use a nested function with nonlocal diameter -> (clean if you define the helper inside the main function)

**Binary Search Tree (BST):-** A binary tree in which the left subtree contains smaller values and the right subtree contains larger values than the root.

AVL Tree:- A self-balancing Binary Search Tree in which the height difference between left and right subtrees is at most 1.

Red-Black Tree:- A self-balancing Binary Search Tree that maintains balance using coloring rules (red and black nodes).

Threaded Binary Tree:- A binary tree in which NULL pointers are replaced with links to the inorder predecessor or successor.
