BINARY SEARCH

```python
def search(nums, target):               # Time Complexity: O(log n), Space Complexity: O(1)
    n = len(nums)
    low,high = 0,n-1
    while low <= high:
        mid = (low + high)//2
        if nums[mid] == target:
            return mid
        elif target < nums[mid]:
            high = mid - 1
        else:
            low = mid + 1
    return -1
# Example usage:
nums = [-1,0,3,5,9,12]
target = 9
print(search(nums, target))  # Output: 4
```

LOWER & UPPER BOUND

lower bound -> smallest index arr[ind]>=x

upper bound -> smallest index arr[ind]>x

```python
def lowerBound(arr,n,x):
    low,high = 0,n - 1
    res = n
    while low <= high:
        mid = (low+high)//2
        if arr[mid] >= x:
            res = mid
            high = mid - 1
        else:
            low = mid + 1
    return res

def upperBound(arr,n,x):
    low,high = 0,n - 1
    res = n
    while low <= high:
        mid = (low+high)//2
        if arr[mid] > x:
            res = mid
            high = mid - 1
        else:
            low = mid + 1
    return res

# Example usage:
arr = [1, 2, 2, 3, 3, 5]
x = 3
print(lowerBound(arr, len(arr), x))  # Output: 6
print(upperBound(arr, len(arr), x))  # Output: 6

```

LOWER & UPPER BOUND -> PYTHON MODULE

```python
import bisect

arr = [1, 2, 4, 4, 5, 6]
x = 4

index = bisect.bisect_left(arr, x) 
print(f"Lower Bound index: {index}") # Output: 2 (the first '4')

index_right = bisect.bisect_right(arr, x)
print(f"Upper Bound index: {index_right}") # Output: 4 (the index of '5')
```

#### FLOOR AND CEIL IN AN ARRAY

**Floor -> largest element in array <= x        Upper Bound**

**Ceil -> smallest element in array >= x        Lower Bound**

```python
def getFloorAndCeil(a, n, x):
    low,high = 0,n - 1
    f_res = c_res = -1
    while low <= high:
        mid = (low+high)//2
        if a[mid] == x:
            return a[mid],a[mid]
        elif a[mid] < x:
            f_res = a[mid]
            low = mid + 1
        else:
            c_res = a[mid]
            high = mid - 1
    return f_res,c_res
# Example usage:
a=[3, 4, 7, 8, 8, 10]  
x = 5
n = 6
print(getFloorAndCeil(a, n, x))  # Output: (4, 7)
```

#### TIP

for l<=r remember we are searching for exact element

for l<r  remember we are searching for boundary ( min , max , first occ. , last occ. , etc)
