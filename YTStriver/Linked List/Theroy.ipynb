### SINGLY LINKED LIST

**CREATION**

```python
class Node:
    def __init__(self,data):
        self.data = data
        self.next = None

# head = None       when we print head, it will give Null.
head = Node(10)
second = Node(20)
third = Node(30)

head.next = second
second.next = third

# Traversal
def traversal(head):
    curr = head
    while curr:
        print(curr.data,end="->")
        curr = curr.next
    print("Null")

traversal(head)
```

**INSERTION**

```python
def insert_begin(head,data):
    new_node = Node(data)
    new_node.next = head
    head = new_node
    return head

def insert_end(head, data):
    new_node = Node(data)

    if head is None:
        return new_node

    current = head

    while current.next:
        current = current.next

    current.next = new_node

    return head

def insert_mid(head,data,position):
    new_node = Node(data)
    if position == 0:
        new_node.next = head
        head = new_node
        return head

    current = head
    for i in range(position - 1):
        if current is None:
            raise IndexError("Position out of bounds")
        current = current.next

    new_node.next = current.next
    current.next = new_node

    return head
```

**DELETION**

```python
def delete_begin(head):
    if head is None:
        return None

    head = head.next
    return head

def delete_end(head):
    if head is None:
        return None

    if head.next is None:
        return None

    current = head
    while current.next.next:
        current = current.next

    current.next = None
    return head

def delete_value(head, key):
    if head is None:
        return None

    if head.data == key:
        return head.next

    current = head

    while current.next:
        if current.next.data == key:
            current.next = current.next.next
            return head
        current = current.next

    return head
```
