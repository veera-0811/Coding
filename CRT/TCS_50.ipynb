CHECK PRIME -> SEIVE of Eratosthenes ALGORITHM

```python
def seive(n):
    prime = [True]*(n+1)
    p = 2
    while p*p <= n:
        if prime[p]:
            for i in range(p*p,n+1,p):
                prime[i] = False
        p += 1

    res = []
    for p in range(2,n+1):
        if prime[p]:
            res.append(p)
    return res

n = int(input("Enter a number:"))
res = seive(n)
print(*res)
```

ANAGRAM

```python
from collections import Counter
class Solution:
    def isAnagram(self, s: str, t: str) -> bool:
        if len(s) != len(t):
            return False
        h1 = Counter(s)
        for ch in t:
            if ch not in s:
                return False
            h1[ch] -= 1
            if h1[ch] == 0:
                del h1[ch]
        return not h1

"""
s1 = input().lower()
s2 = input().lower()

if sorted(s1) == sorted(s2):
    print("Anagram")
else:
    print("Not Anagram")
"""
```

Bubble sort

```python
n = len(arr)
arr = [5,4,3,2,1]
print(arr)
for i in range(n):
    for j in range(n - i - 1):
        if arr[j] > arr[j + 1]:
            arr[j], arr[j + 1] = arr[j + 1], arr[j]
print(*arr)
```

Insertion sort

```python
arr = [5,4,3,2,1]
n = len(arr)
for i in range(1,n):
    j = i
    while arr[j] < arr[j - 1] and j > 0:
        arr[j],arr[j-1] = arr[j-1],arr[j]
        j -= 1
print(arr)
```

Selection sort

```python
n = len(arr)
arr = [64,25,12,22,11]
for i in range(0, n - 1):
    curr_min_index = i
    for j in range(i + 1, n):
        if arr[j] < arr[curr_min_index]:
            curr_min_index = j
    arr[i], arr[curr_min_index] = arr[curr_min_index], arr[i]
print(arr)
```

Merge sort

```python
def merge_sort(arr):
    if len(arr) > 1:

        mid = len(arr) // 2

        left = arr[:mid]
        right = arr[mid:]

        merge_sort(left)
        merge_sort(right)

        i = j = k = 0

        while i < len(left) and j < len(right):
            if left[i] < right[j]:
                arr[k] = left[i]
                i += 1
            else:
                arr[k] = right[j]
                j += 1
            k += 1

        while i < len(left):
            arr[k] = left[i]
            i += 1
            k += 1

        while j < len(right):
            arr[k] = right[j]
            j += 1
            k += 1


arr = [38, 27, 43, 3, 9, 82, 10]

merge_sort(arr)

print(arr)
```

Quick sort

```python
def quicksort(arr,left,right):
    if left < right:
        pivot = partition(arr,left,right)
        quicksort(arr,left,pivot-1)
        quicksort(arr,pivot+1,right)

def partition(arr,left,right):
    pivot = arr[right]
    i = left - 1
    for j in range(left,right):
        if arr[j] <= pivot:
            i += 1
            arr[i],arr[j] = arr[j],arr[i]
    arr[i+1],arr[right] = arr[right],arr[i+1]
    return i + 1

arr = [10, 7, 8, 9, 1, 5]
n = len(arr)
quicksort(arr,0,n-1)
print(arr)
```

```python
rows = int(input("Enter number of rows: "))
cols = int(input("Enter number of columns: "))

print("Enter first matrix:")
A = []
for i in range(rows):
    A.append(list(map(int, input().split())))

print("Enter second matrix:")
B = []
for i in range(rows):
    B.append(list(map(int, input().split())))

C = []

for i in range(rows):
    row = []
    for j in range(cols):
        row.append(A[i][j] + B[i][j])
    C.append(row)

print("Sum of matrices:")
for row in C:
    print(*row)
```

Matrix Multiplication

```python
r1 = int(input("Rows of A: "))
c1 = int(input("Columns of A: "))

print("Enter Matrix A:")
A = [list(map(int, input().split())) for _ in range(r1)]

r2 = int(input("Rows of B: "))
c2 = int(input("Columns of B: "))

print("Enter Matrix B:")
B = [list(map(int, input().split())) for _ in range(r2)]

if c1 != r2:
    print("Matrix multiplication not possible")
else:
    C = [[0] * c2 for _ in range(r1)]

    for i in range(r1):
        for j in range(c2):
            for k in range(c1):
                C[i][j] += A[i][k] * B[k][j]

    print("Product Matrix:")
    for row in C:
        print(*row)
```

Right rotate an array by k places

```python
class Solution:
    def rotate(self, nums: list[int], k: int) -> None:
        # Do not return anything, modify nums in-place instead.
        def reverse(arr,l,r):
            while l<r:
                arr[l],arr[r]=arr[r],arr[l]
                l+=1
                r-=1
        n=len(nums)
        k=k%n
        reverse(nums,n-k,n-1)
        reverse(nums,0,n-k-1)
        reverse(nums,0,n-1)        
```

perfect number

```python
import math
class Solution:
    def checkPerfectNumber(self, num: int) -> bool:
        if num<=1:
            return False
        sum=1
        n=int(math.sqrt(num))
        for i in range(2,n+1):
            if num%i==0:
                sum+=i
                if num//i!=i:
                    sum+=num//i
        return sum==num
```

Strong number

```python
n = int(input())

temp = n
s = 0

while n > 0:
    digit = n % 10

    fact = 1
    for i in range(1, digit + 1):
        fact *= i

    s += fact
    n //= 10

if s == temp:
    print("Strong Number")
else:
    print("Not Strong Number")
```
